```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


import os
csv_files = [f for f in os.listdir('wb_data') if f.startswith('API_SP.POP.TOTL') and f.endswith('.csv')]
file_path = os.path.join('wb_data', csv_files[0])

df = pd.read_csv(file_path, skiprows=4)


countries = ['India', 'United States', 'Indonesia', 'Pakistan', 'Brazil', 'Nigeria', 'Bangladesh', 'Mexico', 'Japan', 'Germany']
df_filtered = df[df['Country Name'].isin(countries)]


df_sorted = df_filtered.sort_values(by='2023', ascending=False)

# -----------------------------------------------------------
# CORRECT VISUALIZATION: Categorical Distribution
# -----------------------------------------------------------
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Plotting the distribution across countries
sns.barplot(
    data=df_sorted,
    x='2023',
    y='Country Name',
    palette='viridis',
    hue='Country Name',
    legend=False
)

# Turn off scientific scientific notation notation (like 1e9)
plt.ticklabel_format(style='plain', axis='x')

plt.title('Distribution of Total Population Across Selected Countries (2023)', fontsize=14, fontweight='bold')
plt.xlabel('Total Population', fontsize=12)
plt.ylabel('Country', fontsize=12)

plt.tight_layout()
plt.show()
```
